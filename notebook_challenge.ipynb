### Batch Normalization layers in AI models
A Batch Normalization (BN) layer normalizes intermediate values in the models, and adjusts their distribution using a learnable scaling vector $\gamma$ and bias $\beta$:

$$y = \gamma \odot \frac{x - \mu}{\sqrt{\sigma^2 + \epsilon}} + \beta$$

In many works, these scaling factors $\gamma$ are used to embed white-box watermarks, as they are very easy to manipulate and robust to popular attacks. But perhaps **they are not that secure against a motivated thief like yourself**.
